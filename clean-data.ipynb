```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_excel('template_misi.xlsx', skiprows=2, header=0)

data.head()
```

```python
data['Nama Produk'] = data['Nama Produk'].str.strip()
data['Nama Produk'] = data['Nama Produk'].str.title()
data['Kategori'] = data['Kategori'].str.strip()
data['Kategori'] = data['Kategori'].str.title()

for col in data.columns:
    if data[col].dtype == 'object':
        data[col] = data[col].str.replace(r'\s+', ' ', regex=True)

data = data.dropna()
data.tail()
```

```python
data['Kode_Number'] = data['Kode Produk Final'].str.extract(r'(\d+)')
data['Kode_Number'] = data['Kode_Number'].astype(int)

data = data.sort_values(by='Kode_Number')
data.head()
```

```python
data = data[data['Kode Produk Final'].str.len() <= 8]
data = data.drop_duplicates(subset='Kode Produk Final', keep='first')
data['No.'] = range(1, len(data) + 1)
data.head()
```

```python
# Assuming 'Harga Awal' is a column in your DataFrame
# Apply the logic and format the 'Harga Awal' column
def format_harga(harga):
    if harga > 1000000:
        harga = harga / 10
    elif harga < 10000:
        harga = harga * 10
    return "Rp {:,.0f}".format(harga)

data['Harga Awal'] = data['Harga Awal'].astype(int).apply(format_harga)
data.head()
data
```

```python
# Delete the 'Kode_Number' column
data = data.drop('Kode_Number', axis=1)

# Export the modified DataFrame to an Excel file
from google.colab import files
data.to_excel('modified_data.xlsx', index=False)
files.download('modified_data.xlsx')
```
